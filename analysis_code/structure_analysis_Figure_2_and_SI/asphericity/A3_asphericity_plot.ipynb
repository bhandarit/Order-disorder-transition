Time series A3 plot

```python
import pandas as pd
import matplotlib.pyplot as plt
import os
import numpy as np

folders = {
    "IM30": "/home/POLY/bhandarit/Desktop/chain_n_chain/aspericity/full_length/",
    "IM30 H0-3": "/home/POLY/bhandarit/Desktop/chain_n_chain/aspericity/H0_3/",
    "IM30 H4-6": "/home/POLY/bhandarit/Desktop/chain_n_chain/aspericity/H4_6/",
}

def plot_A3_timeseries(folder_name, folder_path):

    csv_path = os.path.join(folder_path, "A3_vs_Time.csv")
    print(f"Processing: {csv_path}")

    df = pd.read_csv(csv_path)
    df = df.sort_values(["Bias", "Time"])

    bias_values = sorted(df["Bias"].unique())

    plt.figure(figsize=(10, 7))

    for bias in bias_values:
        sub = df[df["Bias"] == bias]

        mean_series = sub["A3_mean"].reset_index(drop=True)
        std_series  = sub["A3_std"].reset_index(drop=True) / np.sqrt(12)

        time = mean_series.index / 1000.0

        plt.plot(
            time,
            mean_series,
            linewidth=2.5,
            label=f"{bias*100:.0f}%"
        )

        plt.fill_between(
            time,
            mean_series - std_series,
            mean_series + std_series,
            alpha=0.25
        )

    # Labels
    plt.xlabel("Time (μs)", fontsize=30, fontweight="bold")
    plt.ylabel("A3", fontsize=30, fontweight="bold")

    # Title
    plt.title(f"{folder_name}", fontsize=30, fontweight="bold")

    # Ticks
    plt.xticks(fontsize=26, fontweight='bold')
    plt.yticks(fontsize=26, fontweight='bold')

    # Y-axis limits
    plt.ylim(0, 1)

    # Legend
    legend = plt.legend(
        title="H-bond Strength",
        fontsize=20,
        title_fontsize=22
    )
    plt.setp(legend.get_title(), fontweight='bold')
    for text in legend.get_texts():
        text.set_fontweight('bold')

    # Thick border box
    ax = plt.gca()
    for spine in ax.spines.values():
        spine.set_linewidth(2)

    plt.tight_layout()

    save_path = f"A3_timeseries_{folder_name.replace(' ', '_')}.png"
    plt.savefig(save_path, dpi=600)
    print(f"Saved: {save_path}\n")

    plt.close()


for name, path in folders.items():
    plot_A3_timeseries(name, path)
```

Plot A3 vs HBS

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Paths to the folders
folders = [
    "/home/POLY/bhandarit/Desktop/chain_n_chain/aspericity/full_length/",
    "/home/POLY/bhandarit/Desktop/chain_n_chain/aspericity/H0_3/",
    "/home/POLY/bhandarit/Desktop/chain_n_chain/aspericity/H4_6/"
]

# Filename (same in all folders)
filename = "Average_A3_vs_Bias.csv"

# Labels for the legend
labels = ['IM30', 'IM30 H0-3', 'IM30 H4-6']

# Colors for the curves
colors = ['red', 'green', 'blue']

plt.figure(figsize=(10,7))

for folder, label, color in zip(folders, labels, colors):
    filepath = os.path.join(folder, filename)
    if os.path.exists(filepath):
        df = pd.read_csv(filepath)
        df["Bias"] = df["Bias"] * 100  # Convert to %
        df = df.sort_values("Bias")    # Ensure smooth curves
        
        # Error bars
        plt.errorbar(
            df["Bias"],
            df["Average_A3_mean"],
            yerr=df["Average_A3_std"] / np.sqrt(12),  # SEM propagation if 12 samples
            label=label,
            color=color,
            marker="o",
            linestyle="-",
            linewidth=2.5,
            markersize=7,
            capsize=4
        )
    else:
        print(f"File not found: {filepath}")

# Axis labels
plt.xlabel("H-Bond Strength (%)", fontsize=30, fontweight='bold')
plt.ylabel("Asphericity (A3)", fontsize=30, fontweight='bold')

# Ticks
plt.xticks(fontsize=26, fontweight='bold')
plt.yticks(fontsize=26, fontweight='bold')

# Legend with bold text and title
legend = plt.legend( fontsize=26, title_fontsize=22)
plt.setp(legend.get_title(), fontweight='bold')
for text in legend.get_texts():
    text.set_fontweight('bold')

# Closed box
ax = plt.gca()
for spine in ax.spines.values():
    spine.set_linewidth(2)

plt.tight_layout()

# Save figure in current folder
plt.savefig("Average_A3_vs_Bias_all_styled.png", dpi=600, bbox_inches='tight')
plt.show()

print("Figure saved as: Average_A3_vs_Bias_all_styled.png")
```
